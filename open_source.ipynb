```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
```

```python
load_dotenv()
```

```python
client = ChatGroq(
    api_key = os.getenv("GROK_API_KEY"),
    model = "openai/gpt-oss-120b"

)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
messages = [
    SystemMessage(content="Translate the input from english to german"),
    HumanMessage(content="You're good enough")
]
```

```python
results = client.invoke(messages)
results
```

##### Output Passer

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser()
parser.invoke(results)
```

```python
#### Using LCEL
chain = client | parser
chain.invoke(messages)
```

```python
####can decide to use Chat template
from langchain_core.prompts import ChatPromptTemplate
```

```python
generic_prompt = "Translate the text into {language}:"

prompt = ChatPromptTemplate(
    [
        ("system", generic_prompt),
        ("user", "{text}")
    ]
)
```

```python
result = prompt.invoke({"language":"German", "text":"You're good enough"})
```

```python
result.to_messages()
```

```python
#chaining component with LCEL
chain = prompt | client | parser
chain.invoke({"language":"German", "text":"You're good enough"})
```
